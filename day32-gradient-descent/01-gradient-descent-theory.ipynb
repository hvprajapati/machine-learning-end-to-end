## What is Gradient Descent?

- **Gradient Descent is an optimization algorithm** used to find the values of model parameters that minimize a **loss/cost function**.
- In machine learning, the model makes predictions, calculates the error, and Gradient Descent helps us adjust the model parameters to reduce this error.
- The main idea is:

> **Move in the direction that decreases the loss.**

- It is widely used in Linear Regression, Logistic Regression, Neural Networks, and many other ML algorithms.

## Why Do We Need Gradient Descent?

- A machine learning model has parameters such as **weights** and **biases**.
- Different parameter values produce different predictions and therefore different loss values.
- Our goal is to find the parameter values that give the **minimum possible loss**.

```text
Parameters
     ↓
Predictions
     ↓
Loss
     ↓
Adjust Parameters
     ↓
Lower Loss
     ↓
Repeat

## Intuition Behind Gradient Descent

Imagine you are standing somewhere on a mountain and want to reach the **lowest point**.

- You cannot see the entire mountain.
- You can only determine the direction in which the ground slopes.
- You take a step in the direction that goes **downhill**.
- You repeat this process until you reach a low point.

### In Gradient Descent

| Real-World Intuition | Gradient Descent |
|---|---|
| Mountain | Loss / Cost Function |
| Current position | Current parameter values |
| Downhill direction | Negative Gradient |
| Step size | Learning Rate |
| Lowest point | Minimum Loss |

> **Core idea:** Gradient Descent repeatedly moves in the direction of lower loss until it reaches a minimum.

## What is a Gradient?

- A **gradient** tells us the direction and rate of the **steepest increase** of a function.
- Therefore, the **negative gradient** tells us the direction of the **steepest decrease**.
- Since Gradient Descent wants to **minimize the loss**, we move in the direction of the **negative gradient**.

### Intuition

```text
Gradient
   ↓
Direction of increasing loss

Negative Gradient
   ↓
Direction of decreasing loss

Cell 5 — Update Rule
Gradient Descent Update Rule

The general update rule is:

$$ \theta_{new} = \theta_{old} - \alpha \nabla J(\theta) $$

Where:

$\theta$ → Model parameters such as weights and bias
$\alpha$ → Learning rate
$J(\theta)$ → Loss / Cost function
$\nabla J(\theta)$ → Gradient of the loss function
Simple Intuition

New Parameter

$$ = $$

Old Parameter

$$ - $$

Learning Rate × Gradient

The minus sign is important because we want to move in the direction opposite to the gradient, which decreases the loss.

## Learning Rate

The **learning rate ($\alpha$)** controls the size of the step taken during each parameter update.

### If Learning Rate is Too Small

- Very small updates
- Training becomes slow
- May require many iterations to reach the minimum

### If Learning Rate is Too Large

- Very large updates
- May overshoot the minimum
- Loss may oscillate
- In extreme cases, the algorithm may diverge

### Good Learning Rate

- Makes sufficiently large progress
- Moves toward the minimum
- Provides stable convergence

**Too Small**<br>
↓<br>
🐢 Very Slow

**Good**<br>
↓<br>
🎯 Stable Convergence

**Too Large**<br>
↓<br>
↔️ Oscillation / Divergence


## How Does Gradient Descent Work?

Gradient Descent is an **iterative optimization algorithm**.

The basic process is:

1. Initialize the model parameters.
2. Make predictions using the current parameters.
3. Calculate the loss.
4. Calculate the gradient of the loss.
5. Update the parameters.
6. Repeat the process.
7. Stop when the algorithm sufficiently converges.

### Process

```mermaid
graph TD
    A[Initialize Parameters] --> B[Make Predictions]
    B --> C[Calculate Loss]
    C --> D[Calculate Gradient]
    D --> E[Update Parameters]
    E --> F{Has it Converged?}
    F -- No --> B
    F -- Yes --> G[Stop]


## Types of Gradient Descent

Gradient Descent can be classified based on **how much training data is used to calculate the gradient in one update**.

There are three major types:

### 1. Batch Gradient Descent

- Uses the **entire training dataset** to calculate one gradient update.
- Produces relatively smooth and stable updates.
- Can become computationally expensive for very large datasets.

### 2. Stochastic Gradient Descent (SGD)

- Uses **one training example** to calculate each gradient update.
- Performs frequent parameter updates.
- Faster individual updates but produces a noisier optimization path.

### 3. Mini-Batch Gradient Descent

- Uses a **small batch of training examples** for each update.
- Provides a balance between Batch Gradient Descent and SGD.
- Commonly used in modern machine learning and deep learning.

### Overview

```mermaid
graph TD
    GD[Gradient Descent] --> B[Batch]
    GD --> S[SGD]
    GD --> MB[Mini-Batch]
    
    B --> B1[All samples]
    S --> S1[1 sample]
    MB --> MB1[Small batch]


## Batch Gradient Descent

**Batch Gradient Descent** calculates the gradient using the **entire training dataset** before making one parameter update.

### How it works

```mermaid
graph TD
    A[Entire Dataset] --> B[Calculate Gradient]
    B --> C[Update Parameters]
    C --> D[Repeat]


## Stochastic Gradient Descent (SGD)

**Stochastic Gradient Descent** calculates the gradient using **one training example at a time**.

### How it works

```mermaid
graph TD
    A[One Sample] --> B[Calculate Gradient]
    B --> C[Update Parameters]
    C --> D[Next Sample]
    D --> E[Calculate Gradient]
    E --> F[Update Parameters]
    F --> G[Repeat]


## Mini-Batch Gradient Descent

**Mini-Batch Gradient Descent** calculates the gradient using a **small group of training examples**.

For example:

```mermaid
graph TD
    A[Dataset] --> B[Batch 1 &rarr; 32 samples]
    A --> C[Batch 2 &rarr; 32 samples]
    A --> D[Batch 3 &rarr; 32 samples]
    A --> E[...]


## Batch vs SGD vs Mini-Batch

The primary difference is the **number of training examples used for one parameter update**.

| Method | Samples per Update | Optimization Path | Typical Behavior |
|---|---:|---|---|
| Batch GD | Entire dataset | Smooth | Stable but potentially slow per update |
| SGD | 1 | Noisy | Frequent and fast updates |
| Mini-Batch GD | Small batch | Relatively stable | Good balance |

### Easy Way to Remember

- **Batch** $\rightarrow$ ALL samples
- **SGD** $\rightarrow$ ONE sample
- **Mini-Batch** $\rightarrow$ FEW samples


## Gradient Descent in Linear Regression

In Linear Regression, we want to find the values of the model parameters that minimize the **cost function**.

For Simple Linear Regression:

$$
\hat{y} = wx + b
$$

Where:

- $w$ → Weight / slope
- $b$ → Bias / intercept

The model first makes predictions:

$$
\hat{y}=wx+b
$$

Then we calculate the error between the actual and predicted values.

A common cost function is **Mean Squared Error (MSE)**:

$$
J(w,b)=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

Gradient Descent then updates $w$ and $b$ to reduce this cost.

```mermaid
graph TD
    A[Initial w, b] --> B[Make Predictions]
    B --> C[Calculate MSE]
    C --> D[Calculate Gradients]
    D --> E[Update w and b]
    E --> F[Lower MSE]
    F --> G[Repeat]


## Parameter Updates in Linear Regression

For Linear Regression, Gradient Descent updates both the **weight** and **bias**.

### Weight Update

$$
w_{new}=w_{old}-\alpha\frac{\partial J}{\partial w}
$$

### Bias Update

$$
b_{new}=b_{old}-\alpha\frac{\partial J}{\partial b}
$$

Where:

- $w$ → Weight
- $b$ → Bias
- $\alpha$ → Learning rate
- $J$ → Cost function

### Important Idea

The parameters are updated repeatedly until the cost becomes sufficiently small or the optimization converges.

```mermaid
graph TD
    A[w, b] --> B[Prediction]
    B --> C[Cost]
    C --> D[Gradient]
    D --> E[New w, b]
    E --> F[Repeat]


## Minimum of the Cost Function

The objective of Gradient Descent is to reach a point where the cost is minimized.

For a simple convex cost function such as the standard MSE objective of Linear Regression, there is a **global minimum** under the usual conditions.

<pre>
Cost
 │
 │\
 │ \
 │  \
 │   \       /
 │    \_____/
 │       ↑
 │    Minimum
 └──────────────── Parameter
</pre>


## Convergence

**Convergence** means that the optimization process has reached a point where further parameter updates produce very little improvement.

A model may be considered converged when:

- The change in loss becomes very small.
- The change in parameters becomes very small.
- A predefined number of iterations has been reached.

### Example

- **Iteration 1** $\rightarrow$ Loss = 1000
- **Iteration 2** $\rightarrow$ Loss = 700
- **Iteration 3** $\rightarrow$ Loss = 520
- **Iteration 4** $\rightarrow$ Loss = 510
- **Iteration 5** $\rightarrow$ Loss = 509.9
- **Iteration 6** $\rightarrow$ Loss = 509.89 &nbsp; $\leftarrow$ *(Very small improvement)*


## Gradient Descent — Final Summary

- **Gradient Descent** is an optimization algorithm.
- Its goal is to **minimize a loss/cost function**.
- The **gradient** points toward increasing loss.
- Therefore, Gradient Descent moves in the direction of the **negative gradient**.
- The **learning rate** controls the size of each update.
- The process is repeated iteratively until convergence.

### Three Main Types

- **Batch GD** &rarr; Entire dataset per update
- **SGD** &rarr; One sample per update
- **Mini-Batch GD** &rarr; Small batch per update
